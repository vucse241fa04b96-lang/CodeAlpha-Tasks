## Data Visualization

Data Visualization is the process of representing raw data using charts, graphs, and dashboards.

In this task, the dataset created during Task 1 is visualized using Python libraries such as Matplotlib and Seaborn.

The visualizations help to understand book prices, ratings, availability, and relationships between different variables.

## Loading the Dataset

The dataset created during Task 1 is loaded using Pandas.

```python
import pandas as pd

df = pd.read_csv("CodeAlpha_Task1_Books_Dataset.csv")

print("Dataset loaded successfully!")
```

## Libraries Used

- Pandas – Data handling
- NumPy – Numerical operations
- Matplotlib – Creating charts and graphs
- Seaborn – Creating attractive statistical visualizations

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("All libraries imported successfully.")
```

## Viewing the Dataset

The first few records of the dataset are displayed.

```python
df.head()
```

## Understanding the Dataset

The dataset shape, column names, and data types are examined before creating visualizations.

```python
print("Number of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)
```

## Data Cleaning

The Price and Rating columns are converted into numerical values so that they can be used for visualization.

```python
df["Price"] = pd.to_numeric(df["Price"], errors="coerce")
df["Rating"] = pd.to_numeric(df["Rating"], errors="coerce")

df = df.dropna(subset=["Price", "Rating"])

print("Data cleaning completed!")
```

## Basic Statistical Summary

The basic statistics of the numerical columns are calculated.

```python
df[["Price", "Rating"]].describe()
```

## Questions for Data Visualization

The following questions are considered before creating charts:

1. What is the distribution of book prices?
2. Which rating is most common?
3. How many books are available?
4. What is the average price for each rating?
5. Is there a relationship between price and rating?
6. Which books are the most expensive?

## Distribution of Book Prices

A histogram is used to understand how book prices are distributed.

It helps identify whether most books are low-priced, medium-priced, or high-priced.

```python
plt.figure(figsize=(10, 6))

plt.hist(df["Price"], bins=20, edgecolor="black")

plt.xlabel("Book Price")
plt.ylabel("Number of Books")
plt.title("Distribution of Book Prices")

plt.show()
```

## Boxplot of Book Prices

A boxplot is used to identify the spread of book prices and possible outliers.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("CodeAlpha_Task1_Books_Dataset.csv")
df["Price"] = pd.to_numeric(df["Price"], errors="coerce")
df["Rating"] = pd.to_numeric(df["Rating"], errors="coerce")
df = df.dropna(subset=["Price", "Rating"])

plt.figure(figsize=(10, 5))

sns.boxplot(x=df["Price"])

plt.xlabel("Book Price")
plt.title("Boxplot of Book Prices")

plt.show()
```

## Distribution of Book Ratings

A bar chart is used to display the number of books for each rating.

```python
rating_counts = df["Rating"].value_counts().sort_index()

plt.figure(figsize=(8, 5))

plt.bar(
    rating_counts.index,
    rating_counts.values,
    edgecolor="black"
)

plt.xlabel("Rating")
plt.ylabel("Number of Books")
plt.title("Distribution of Book Ratings")

plt.show()
```

## Percentage Distribution of Ratings

A pie chart is used to show the percentage contribution of each rating category.

```python
plt.figure(figsize=(8, 8))

plt.pie(
    rating_counts.values,
    labels=rating_counts.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Percentage Distribution of Book Ratings")

plt.show()
```

## Average Price by Rating

The average price for each rating is calculated and visualized using a bar chart.

```python
average_price_by_rating = (
    df.groupby("Rating")["Price"]
    .mean()
    .round(2)
)

print(average_price_by_rating)
```

```python
plt.figure(figsize=(8, 5))

plt.bar(
    average_price_by_rating.index,
    average_price_by_rating.values,
    edgecolor="black"
)

plt.xlabel("Rating")
plt.ylabel("Average Price")
plt.title("Average Book Price by Rating")

plt.show()
```

## Relationship Between Price and Rating

A scatter plot is used to visualize the relationship between book price and rating.

```python
plt.figure(figsize=(10, 6))

plt.scatter(
    df["Rating"],
    df["Price"],
    alpha=0.7
)

plt.xlabel("Rating")
plt.ylabel("Book Price")
plt.title("Relationship Between Book Price and Rating")

plt.show()
```

## Correlation Heatmap

A heatmap is used to display the correlation between numerical variables.

The correlation value ranges from -1 to +1.

```python
correlation_matrix = df[["Price", "Rating"]].corr()

plt.figure(figsize=(7, 5))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    linewidths=0.5
)

plt.title("Correlation Heatmap")

plt.show()
```

## Top 10 Most Expensive Books

The ten most expensive books are identified and displayed using a bar chart.

```python
top_expensive_books = df.sort_values(
    by="Price",
    ascending=False
).head(10)

top_expensive_books[["Title", "Price"]]
```

```python
plt.figure(figsize=(12, 6))

plt.barh(
    top_expensive_books["Title"],
    top_expensive_books["Price"]
)

plt.xlabel("Book Price")
plt.ylabel("Book Title")
plt.title("Top 10 Most Expensive Books")

plt.gca().invert_yaxis()

plt.show()
```

## Book Availability Analysis

The availability of books is analyzed to understand how many books are available or unavailable.

```python
availability_counts = df["Availability"].value_counts()

print(availability_counts)
```

```python
plt.figure(figsize=(8, 5))

plt.bar(
    availability_counts.index,
    availability_counts.values,
    edgecolor="black"
)

plt.xlabel("Availability")
plt.ylabel("Number of Books")
plt.title("Book Availability Distribution")

plt.xticks(rotation=45)

plt.show()
```

## Price Distribution for Each Rating

A boxplot is used to compare the price distribution of books belonging to different rating categories.

```python
plt.figure(figsize=(10, 6))

sns.boxplot(
    x="Rating",
    y="Price",
    data=df
)

plt.xlabel("Rating")
plt.ylabel("Book Price")
plt.title("Book Price Distribution by Rating")

plt.show()
```

## Count of Books by Rating

A Seaborn countplot is used to create an attractive visualization of rating counts.

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    x="Rating",
    data=df
)

plt.xlabel("Rating")
plt.ylabel("Number of Books")
plt.title("Number of Books by Rating")

plt.show()
```

## Pairplot of Numerical Variables

A pairplot is used to visualize relationships and distributions between numerical variables.

```python
sns.pairplot(
    df[["Price", "Rating"]]
)

plt.show()
```

## Simple Data Visualization Dashboard

A dashboard combines multiple visualizations into one figure to provide a quick overview of the dataset.

```python
fig, axes = plt.subplots(2, 2, figsize=(16, 10))

# Histogram
axes[0, 0].hist(df["Price"], bins=20, edgecolor="black")
axes[0, 0].set_title("Distribution of Book Prices")
axes[0, 0].set_xlabel("Price")
axes[0, 0].set_ylabel("Number of Books")

# Rating count
axes[0, 1].bar(
    rating_counts.index,
    rating_counts.values,
    edgecolor="black"
)
axes[0, 1].set_title("Distribution of Ratings")
axes[0, 1].set_xlabel("Rating")
axes[0, 1].set_ylabel("Number of Books")

# Scatter plot
axes[1, 0].scatter(
    df["Rating"],
    df["Price"],
    alpha=0.7
)
axes[1, 0].set_title("Price vs Rating")
axes[1, 0].set_xlabel("Rating")
axes[1, 0].set_ylabel("Price")

# Average price by rating
axes[1, 1].bar(
    average_price_by_rating.index,
    average_price_by_rating.values,
    edgecolor="black"
)
axes[1, 1].set_title("Average Price by Rating")
axes[1, 1].set_xlabel("Rating")
axes[1, 1].set_ylabel("Average Price")

plt.suptitle("Book Dataset Visualization Dashboard", fontsize=16)

plt.tight_layout()

plt.show()
```

## Saving the Visualization

The dashboard is saved as an image file for documentation and portfolio purposes.

```python
fig, axes = plt.subplots(2, 2, figsize=(16, 10))

axes[0, 0].hist(df["Price"], bins=20, edgecolor="black")
axes[0, 0].set_title("Distribution of Book Prices")
axes[0, 0].set_xlabel("Price")
axes[0, 0].set_ylabel("Number of Books")

axes[0, 1].bar(
    rating_counts.index,
    rating_counts.values,
    edgecolor="black"
)
axes[0, 1].set_title("Distribution of Ratings")
axes[0, 1].set_xlabel("Rating")
axes[0, 1].set_ylabel("Number of Books")

axes[1, 0].scatter(
    df["Rating"],
    df["Price"],
    alpha=0.7
)
axes[1, 0].set_title("Price vs Rating")
axes[1, 0].set_xlabel("Rating")
axes[1, 0].set_ylabel("Price")

axes[1, 1].bar(
    average_price_by_rating.index,
    average_price_by_rating.values,
    edgecolor="black"
)
axes[1, 1].set_title("Average Price by Rating")
axes[1, 1].set_xlabel("Rating")
axes[1, 1].set_ylabel("Average Price")

plt.suptitle("Book Dataset Visualization Dashboard", fontsize=16)

plt.tight_layout()

plt.savefig("CodeAlpha_Task3_Data_Visualization.png", dpi=300)

plt.show()

print("Dashboard saved successfully!")
```
